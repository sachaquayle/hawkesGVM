# P-values for asymptotic normality

## 100 realisations

```python
####### Define parameters #######

d = 2
alpha = np.array([[0.2, 0.0],[-0.6,1.2]])
beta = np.array([3.5,2.0])
mu = np.array([0.7,1.0])
alpha_tilde = alpha

####### Simulate data #######

nb_realisations = 1000
size = 5000

hawkes = ExponentialHawkesGVM(model='gvm', mu=mu, alpha=alpha, beta=beta, alpha_tilde=alpha)
times = hawkes.simulate(size=size, nb_realisations=nb_realisations)
```

```python
hawkes.fit(times)
```

```python
fig,ax = plt.subplots(2,2)
for i in range(2):
    for j in range(2):
        estimations = [e[1][i,j] for e in hawkes.multiple_estimations]
        ax[i,j].hist(estimations-alpha[i,j], edgecolor='white', bins=15)
        ax[i,j].set_title(fr'$\alpha_{{{i+1},{j+1}}}$')

plt.tight_layout()
plt.show()
        
```

```python
p_values = np.zeros((d,d))

for i in range(d):
    for j in range(d):
        estimations = [ est[1][i,j] for est in hawkes.multiple_estimations]
        mean = np.mean(estimations)
        std = np.std(estimations)
        data = (estimations-mean)/std
        p_values[i,j] = kstest(data,'norm').pvalue

p_values
```

## P-values for detection of absence of interactions 

```python
####### Define parameters #######

d = 2
alpha = np.array([[0.2, 0.0],[-0.6,1.2]])
beta = np.array([3.5,2.0])
mu = np.array([0.7,1.0])
alpha_tilde = alpha

####### Simulate data #######

nb_realisations = 25
size = 5000

hawkes = ExponentialHawkesGVM(model='gvm', mu=mu, alpha=alpha, beta=beta, alpha_tilde=alpha)
times = hawkes.simulate(size=size, nb_realisations=nb_realisations)
```

### Under (GVM)

```python
# Step 1

hawkes = ExponentialHawkesGVM()
hawkes.fit(times)

num_tests = hawkes.d*hawkes.d
p_values = np.zeros((hawkes.d,hawkes.d))

for i in range(hawkes.d):
    for j in range(hawkes.d):
        data = np.zeros((hawkes.nb_realisations, 2))
        for l in range(hawkes.nb_realisations):
            data[l,0] = hawkes.multiple_estimations[l][1][i,j]
            data[l,1] = hawkes.multiple_estimations[l][3][i,j]
            
        mean_vector = np.mean(data, axis=0)
        covariance_matrix = np.cov(data, rowvar=False, ddof=1)
        covariance_matrix_inv = np.linalg.inv(covariance_matrix)
        t2_stat = hawkes.nb_realisations * mean_vector.T @ covariance_matrix_inv @ mean_vector
        f_stat = (t2_stat * (hawkes.nb_realisations - 2)) / (2 * (hawkes.nb_realisations - 1))
        
        # Compute the p-value
        p_values[i,j] = 1 - stats.f.cdf(f_stat, 2, hawkes.nb_realisations - 2)

print('p-values for Step 2:', p_values)
```

```python
# Step 3

hawkes.test_absence_interactions() 
hawkes.fit(times)

and_matrix = np.logical_and(hawkes.alpha_zero_coefficients, hawkes.alpha_tilde_zero_coefficients) # positions where both coeffs are zero
num_params = (~and_matrix).sum()
            
df = hawkes.nb_realisations - 1 
estimations_alpha_tilde = np.array( [e[3][~and_matrix] for e in hawkes.multiple_estimations] ).T
estimations_alpha = np.array( [e[1][~and_matrix] for e in hawkes.multiple_estimations] ).T

test_values_hp = estimations_alpha_tilde - estimations_alpha
means_hp = np.mean(test_values_hp, axis=1)  
std_errors_hp = np.std(test_values_hp, axis=1, ddof=1) / np.sqrt(hawkes.nb_realisations)  
t_stats_hp = means_hp / std_errors_hp
p_values_hp = 2 * (1 - stats.t.cdf(np.abs(t_stats_hp), df=df))  
print('p-values for test (HP) in Step 4 (on coefficients where an interaction was detected in Step 2):', p_values_hp)

nb_alpha_tilde_params = (~hawkes.alpha_tilde_zero_coefficients).sum()

df = hawkes.nb_realisations - 1 
estimations_alpha_tilde = np.array( [e[3][~hawkes.alpha_tilde_zero_coefficients] for e in hawkes.multiple_estimations] ).T

means = np.mean(estimations_alpha_tilde, axis=1)  
std_errors = np.std(estimations_alpha_tilde, axis=1, ddof=1) / np.sqrt(hawkes.nb_realisations)  
t_stats = means / std_errors
pvalues = 2 * (1 - stats.t.cdf(np.abs(t_stats), df=df))  

print('\np-values for test (VM) in Step 4 (on coefficients where an interaction was detected in Step 2):', pvalues)
```
